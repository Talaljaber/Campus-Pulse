# 05 · Fairness by language
Macro-F1 per slice (EN, AR, Arabizi) with row and group counts (`fairness.json`). No significance claims: the effective n is about 30 groups (LD-32, LD-33).

```python
import json
from pathlib import Path

import pandas as pd

R = Path("../reports")  # the only directory read (scripts/run_notebooks.py --check-reads-only)


def load(name):
    return json.loads((R / name).read_text(encoding="utf-8"))
f = load("fairness.json")
s = pd.DataFrame(f["slices"])
display(s.pivot_table(index=["dataset", "target"], columns="slice", values="macro_f1").round(3))
display(s.pivot_table(index=["dataset", "target"], columns="slice", values=["n_rows", "n_groups"]))
print("max gap:", f["max_gap"])
```

```python
abl = load("ablations.json")["arms"]
before_after = {arm: abl[arm]["our_test"]["category"]["macro_f1"] for arm in abl
                if arm in ("full", "-augmentation", "+unreviewed_augmentation", "-m2proto")}
print("our_test category macro-F1 by arm (the 'before/after mitigation' view):", before_after)
```
